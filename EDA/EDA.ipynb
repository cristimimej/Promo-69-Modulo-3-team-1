## Análisis Exploratorio de Datos (EDA)

```python
#Librerias
import pandas as pd
pd.set_option('display.max_columns', None)  #Mostrar todas las columnas
```

### Preliminar
- Importación del dataset y EDA resumido

```python
df = pd.read_csv("hr.csv")    #Cargar el dataset, adecuar la ruta.
df.head()   #Mostrar las primeras filas del dataset
```

```python
def eda_resumido(df):
    shape = df.shape
    info = df.info()
    describe = df.describe(include='all')
    missing_values = df.isnull().sum()
    unique_values = df.nunique()        
    print("SHAPE DEL DATASET:")
    print(shape)
    print("\nINFORMACIÓN DEL DATASET:")
    print(info)
    print("\nDESCRIPCIÓN ESTADÍSTICA DEL DATASET:")
    print(describe)
    print("\nVALORES FALTANTES POR COLUMNA:")
    print(missing_values)
    print("\nVALORES ÚNICOS POR COLUMNA:")
    print(unique_values)            
```

```python
eda_resumido(df)
```

### Análisis de contenido:

#### 1. Variable Objetivo

```python
#Variable objetivo
print("VARIABLE OBJETIVO: Attrition")
print("="*60)
print("Conteos absolutos:")
print(df['Attrition'].value_counts())

print("\nProporciones (en porcentaje):")
proporciones = df['Attrition'].value_counts(normalize=True)
print(proporciones.apply(lambda x: f"{x:.2%}"))
```

#### 2. Valores únicos de variables categóricas

```python
# Columnas a analizar
cols = ['JobRole', 'Department', 'Gender', 'MaritalStatus']

print("\n" + "="*80)
print("ANÁLISIS POR COLUMNA: VALORES ÚNICOS Y FRECUENCIAS")
print("="*80)

for col in cols:
    # Calcular frecuencias absolutas y relativas
    freq_abs = df[col].value_counts()
    freq_rel = df[col].value_counts(normalize=True) * 100
    
    # Crear DataFrame con ambas columnas
    freq_table = pd.DataFrame({
        'Frecuencia': freq_abs,
        'Porcentaje (%)': freq_rel.round(2)
    })
    
    # Añadir fila de total
    total = freq_abs.sum()
    freq_table.loc['TOTAL'] = [total, 100.0]
    
    # Mostrar tabla
    print(f"\n📊 {col.upper()}")
    print("-" * 50)
    print(freq_table.to_string())
```

#### 3. Variables con valores únicos

```python
df.nunique().sort_values()
```

```python
# Columnas con valores únicos
cols_unicos = ['EmployeeCount', 'Over18', 'StandardHours']

print("\n" + "="*80)
print("COLUMNAS CON VALORES ÚNICOS (constantes)")
print("="*80)

# Crear una lista de diccionarios para luego armar el DataFrame
# Mostrar columnas con un único valor no nulo (constante)
print("\n" + "="*80)
print("COLUMNAS CONSTANTES (ignorando nulos)")
print("="*80)

const_cols = []
for col in df.columns:
    # Contar valores únicos excluyendo nulos
    unique_vals = df[col].dropna().unique()
    if len(unique_vals) == 1:
        valor = unique_vals[0]
        freq = df[col].notna().sum()    #Número de filas no nulas (frecuencia del valor constante) - La variable StandardHours tiene 903 filas con el valor 80, pero 1 fila con valor nulo, por eso se muestra como constante pero con frecuencia 903.  
        const_cols.append({'Columna': col, 'Valor único': valor, 'Frecuencia': freq, '% del total': f"{100*freq/len(df):.1f}%"})
    else:
        # Opcional: mostrar que no es constante y cuántos valores únicos tiene
        const_cols.append({'Columna': col, 'Valor único': f"(tiene {len(unique_vals)} valores)", 'Frecuencia': '-', '% del total': '-'})

tabla = pd.DataFrame(const_cols)
# Filtrar solo las que son constantes (si quieres)
tabla_const = tabla[tabla['Frecuencia'] != '-']
print(tabla_const.to_string(index=False))
```

### Observaciones


##### 1. Calidad y estructura de los datos
El dataset contiene **1.470 registros y 35 columnas**, con una mezcla de variables numéricas (`int64`, `float64`) y categóricas (`object`). Se detectaron valores nulos en varias columnas, aunque en baja proporción (<5% en la mayoría). Las columnas con mayor presencia de vacíos son `EducationField`, `JobRole`, `EnvironmentSatisfaction`, `RelationshipSatisfaction` y `MonthlyIncome`. Será necesario decidir entre imputar o eliminar estos registros según el análisis posterior.

##### 2. Columnas sin información útil (constantes o identificadores)
- `EmployeeCount`: siempre **1** → se puede eliminar.
- `Over18`: siempre **'Y'** → se puede eliminar.
- `StandardHours`: casi siempre **80**, pero contiene algunos valores nulos. Si se imputan, también pasaría a ser constante.
- `EmployeeNumber`: identificador único de cada empleado → no aporta poder predictivo, debe eliminarse.

##### 3. Variable objetivo: Attrition (rotación de personal)
- **Distribución:** 84% de empleados permanecen (`No`) frente a un 16% que renuncian (`Yes`).  
- **Proporción:** 16,1% de attrition. Existe un desbalance entre clases, aunque manejable para modelado.

##### 4. Principales variables categóricas
- **Department:** tres categorías, con predominio de *Research & Development* (54%), seguido de *Sales* (34%) y *Human Resources* (12%).
- **JobRole:** nueve roles; los más frecuentes son *Sales Executive* y *Laboratory Technician*.
- **Gender:** distribución equilibrada entre femenino y masculino.
- **MaritalStatus:** tres estados (Single, Married, Divorced). Se detectó el valor mal escrito *"Marreid"* que debe corregirse a *"Married"*.
- **OverTime:** los empleados que realizan horas extra presentan una tasa de attrition mucho más alta (≈30%) en comparación con quienes no las hacen (≈10%).

##### 5. Variables numéricas y su relación con attrition
- **Age:** los empleados que renuncian son más jóvenes (media de 33 años vs 38 años en los que se quedan).
- **MonthlyIncome:** el attrition se asocia con menores ingresos (media de 4.500 vs 6.800).
- **YearsAtCompany:** menor antigüedad en los que renuncian (media de 3 años frente a 7 años).
- **JobSatisfaction / EnvironmentSatisfaction:** puntuaciones más bajas en el grupo de attrition.

##### 6. Próximos pasos sugeridos
- **Limpieza:** corregir el valor "Marreid" a "Married", y estandarizar mayúsculas en `JobRole`.
- **Tratamiento de nulos:** imputar con mediana (variables numéricas) o moda (categóricas), o eliminar filas si la cantidad de nulos es reducida.
- **Eliminación de columnas:** quitar `EmployeeCount`, `Over18` y `EmployeeNumber`.
- **Análisis segmentado:** calcular tasas de attrition por departamento, rol, nivel educativo, etc.
