# FASE 2

```python
# SIEMPRE lo primero — hacer una copia del dataset original
# Nunca tocamos el df original
df_clean = df.copy()

print("Copia creada!")
print("Filas   :", df_clean.shape[0])
print("Columnas:", df_clean.shape[1])
```

¿Por qué df.copy()?
Sin .copy() cualquier cambio en df_clean también cambiaría df — Python compartiría la misma tabla en memoria. Con .copy() son dos tablas completamente independientes 😊

```python
# TAREA A1 — Normalizar el texto de JobRole

# ── Antes de limpiar — ver cómo está ─────────────
print("ANTES de limpiar:")
print(df_clean["JobRole"].unique())

# ── La función de limpieza ────────────────────────
def fix_job_role(df):
    # Quitar espacios del principio y del final
    df["JobRole"] = df["JobRole"].str.strip()
    # Poner la primera letra de cada palabra en mayúscula
    df["JobRole"] = df["JobRole"].str.title()
    return df

# ── Aplicar la función ────────────────────────────
df_clean = fix_job_role(df_clean)

# ── Después de limpiar — comprobar resultado ──────
print("\nDESPUES de limpiar:")
print(df_clean["JobRole"].unique())
```

Qué hace cada línea:
def fix_job_role(df): → define una función que recibe un DataFrame y lo devuelve limpio. En Fase 2 todo va en funciones — así lo explica vuestro documento 😊
df["JobRole"].str.strip() → elimina los espacios del principio y del final de cada valor:

' sALES eXECUTIVE ' → 'sALES eXECUTIVE'

df["JobRole"].str.title() → pone la primera letra de cada palabra en mayúscula y el resto en minúscula:

'sALES eXECUTIVE' → 'Sales Executive'

return df → devuelve el DataFrame ya limpio — importante no olvidarlo o la función no sirve de nada 😊
df_clean = fix_job_role(df_clean) → aplica la función sobre nuestra copia y guarda el resultado

### Tarea A1 — JobRole limpio ✅
- Problema: espacios al principio/final + mayúsculas caóticas
- Solución: .str.strip() + .str.title()
- Resultado: 9 roles con formato correcto

```python
# TAREA A2 — Convertir columnas de float a int

# ── Antes — ver los tipos actuales ───────────────
print("ANTES de convertir:")
cols_convertir = ["Age", "JobSatisfaction", "MonthlyIncome", "YearsWithCurrManager"]
for col in cols_convertir:
    print(f"  {col}: {df_clean[col].dtype}")

# ── La función de conversión ──────────────────────
def fix_types(df):
    cols = ["Age", "JobSatisfaction", "MonthlyIncome", "YearsWithCurrManager"]
    for col in cols:
        # Primero rellenamos nulos con 0 para poder convertir
        df[col] = df[col].fillna(0)
        # Luego convertimos a entero
        df[col] = df[col].astype(int)
    return df

# ── Aplicar la función ────────────────────────────
df_clean = fix_types(df_clean)

# ── Después — comprobar que cambió ───────────────
print("\nDESPUES de convertir:")
for col in cols_convertir:
    print(f"  {col}: {df_clean[col].dtype}")

# ── Ver que ya no hay decimales ───────────────────
print("\nPrimeros valores de Age:")
print(df_clean["Age"].head())
```

Qué hace cada línea:
cols = ["Age", "JobSatisfaction", "MonthlyIncome", "YearsWithCurrManager"] → lista de columnas que hay que convertir — todas son float64 pero deberían ser int
for col in cols: → recorre cada columna de la lista
df[col].fillna(0) → antes de convertir hay que rellenar los nulos — porque Python no puede convertir NaN a entero directamente. Ponemos 0 de momento — los nulos los trataremos bien en la Tarea A3
df[col].astype(int) → convierte el tipo de dato:

41.0 → 41
3.0 → 3
4907.0 → 4907

df_clean[col].dtype → muestra el tipo actual de esa columna — así vemos el antes y el después claramente

⚠️ Importante — después de esta celda Age ya no tendrá decimales. Los 0 que pusimos en los nulos los arreglaremos en la siguiente tarea 😊

### Tarea A2 — Tipos de datos corregidos ✅
- Problema: Age, JobSatisfaction, MonthlyIncome,
  YearsWithCurrManager eran float64
- Solución: .fillna(0) + .astype(int)
- Resultado: las 4 columnas son ahora int64
- Nota: los 0 de los nulos se tratarán en Tarea A3

```python
# TAREA A3 — Tratar los valores nulos

# ── Antes — ver cuántos nulos quedan ─────────────
print("ANTES de tratar nulos:")
nulos_antes = df_clean.isnull().sum()
print(nulos_antes[nulos_antes > 0])

# ── La función ────────────────────────────────────
def fix_nulls(df):

    # Columnas NUMÉRICAS → rellenar con la MEDIANA
    cols_mediana = [
        "YearsWithCurrManager",
        "TrainingTimesLastYear",
        "Age",
        "JobSatisfaction",
        "MonthlyIncome"
    ]
    for col in cols_mediana:
        mediana = df[col].median()
        df[col] = df[col].replace(0, mediana)
        print(f"  {col}: nulos rellenados con mediana ({mediana})")

    # Columnas de TEXTO → rellenar con la MODA
    cols_moda = [
        "MaritalStatus",
        "BusinessTravel",
        "EducationField",
        "OverTime",
        "Department"
    ]
    for col in cols_moda:
        moda = df[col].mode()[0]
        df[col] = df[col].fillna(moda)
        print(f"  {col}: nulos rellenados con moda ('{moda}')")

    return df

# ── Aplicar la función ────────────────────────────
print("\nRellenando nulos:")
df_clean = fix_nulls(df_clean)

# ── Después — comprobar que no quedan nulos ───────
print("\nDESPUES de tratar nulos:")
nulos_despues = df_clean.isnull().sum()
nulos_despues = nulos_despues[nulos_despues > 0]
if len(nulos_despues) == 0:
    print("  No quedan nulos!")
else:
    print(nulos_despues)
```

Qué hace cada línea:
cols_mediana = [...] → lista de columnas numéricas con nulos — usamos la mediana porque no se ve afectada por valores extremos 😊
df[col].median() → calcula la mediana de esa columna — el valor del medio
df[col].replace(0, mediana) → reemplaza los 0 que pusimos antes en la Tarea A2 por la mediana real. Usamos replace(0, mediana) en lugar de fillna() porque en la Tarea A2 ya convertimos los nulos a 0
cols_moda = [...] → lista de columnas de texto con nulos — usamos la moda porque no puedes calcular una media de palabras 😊
df[col].mode()[0] → calcula la moda — el valor más frecuente. El [0] es porque mode() devuelve una lista y queremos solo el primero
df[col].fillna(moda) → rellena los nulos con ese valor más frecuente
if len(nulos_despues) == 0: → comprueba si quedan nulos — si la lista está vacía significa que los hemos tratado todos ✅

¿Qué ha pasado?

StandardHours → tiene 164 nulos pero no la tratamos porque la vamos a eliminar en la Tarea B1 de vuestra compañera — es una columna inútil 🗑️
TrainingTimesLastYear → tiene 88 nulos y sí debería haberse rellenado ⚠️

El problema de TrainingTimesLastYear es que en la Tarea A2 no la convertimos a int — se quedó como float64 con nulos reales, no con ceros. Por eso replace(0, mediana) no funcionó con ella 😊

```python
# CORRECCION — TrainingTimesLastYear quedó con nulos
# porque es float y no pasó por fix_types

# Ver el tipo actual
print("Tipo actual:", df_clean["TrainingTimesLastYear"].dtype)

# Rellenar nulos con la mediana
mediana_training = df_clean["TrainingTimesLastYear"].median()
print("Mediana:", mediana_training)

df_clean["TrainingTimesLastYear"] = df_clean["TrainingTimesLastYear"].fillna(mediana_training)

# Convertir a int
df_clean["TrainingTimesLastYear"] = df_clean["TrainingTimesLastYear"].astype(int)

# Comprobar
print("\nNulos que quedan en TrainingTimesLastYear:")
print(df_clean["TrainingTimesLastYear"].isnull().sum())
print("Tipo ahora:", df_clean["TrainingTimesLastYear"].dtype)
```

Qué hace:
df_clean["TrainingTimesLastYear"].median() → calcula la mediana de los valores que sí existen — ignorando los nulos
fillna(mediana_training) → rellena los nulos con esa mediana — aquí usamos fillna directamente porque la columna sigue siendo float con nulos reales, no con ceros
.astype(int) → la convierte a entero ya sin nulos

StandardHours la dejamos con nulos — la eliminará la Pareja B en su Tarea B1 😊
