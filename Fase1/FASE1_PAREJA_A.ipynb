```python
# TAREA A2 — ¿Hay huecos vacíos?

# ── 1. Contar nulos por columna ──────────────────
print("NULOS POR COLUMNA:")
nulos = df.isnull().sum()
print(nulos[nulos > 0].sort_values(ascending=False))

# ── 2. Porcentaje de nulos por columna ───────────
print("\nPORCENTAJE DE NULOS:")
porcentaje = df.isnull().mean() * 100
porcentaje = porcentaje[porcentaje > 0].round(2).sort_values(ascending=False)
print(porcentaje)
```

Qué hace cada línea:
df.isnull().sum() → cuenta cuántos nulos hay en cada columna

df.isnull() → crea una tabla de True/False — True donde hay nulo
.sum() → suma los True de cada columna → total de nulos

nulos[nulos > 0] → filtra y se queda solo con las columnas que tienen algún nulo — las que tienen 0 no nos interesan
.sort_values(ascending=False) → ordena de mayor a menor — las más problemáticas primero
df.isnull().mean() * 100 → en lugar de contar, calcula la proporción de nulos:

.mean() → calcula la media de True/False → da un número entre 0 y 1
* 100 → lo convierte en porcentaje

11 columnas tienen nulos de las 35 totales. Organizados por gravedad:

```python
# TAREA A3 — ¿Hay filas repetidas?

# ── 1. Contar filas duplicadas ───────────────────
duplicados = df.duplicated().sum()
print("Filas duplicadas:", duplicados)

# ── 2. Ver cómo son esas filas duplicadas ────────
if duplicados > 0:
    print("\nAsí son las filas duplicadas:")
    display(df[df.duplicated(keep=False)])
```

Qué hace cada línea:
df.duplicated() → marca como True las filas exactamente iguales a otra anterior
.sum() → cuenta cuántas hay
if duplicados > 0: → solo ejecuta lo siguiente si hay duplicados — si no los hay no tiene sentido buscarlos
df.duplicated(keep=False) → esto es nuevo respecto a antes 😊

Sin keep=False → solo marca la segunda copia como duplicada
Con keep=False → marca todas las copias — así vemos tanto el original como la copia duplicada

display(df[df.duplicated(keep=False)]) → muestra todas las filas duplicadas en formato tabla para poder verlas

¿Qué nos dice este resultado?

```python
# TAREA A4 — Estadísticas de columnas numéricas

# ── 1. Estadísticas generales ────────────────────
print("ESTADISTICAS COLUMNAS NUMERICAS:")
display(df.describe().T)

# ── 2. Columnas más relevantes para el proyecto ──
print("\nCOLUMNAS CLAVE — detalle:")

columnas_clave = ["Age", "MonthlyIncome",
                  "YearsAtCompany", "TotalWorkingYears",
                  "JobSatisfaction", "WorkLifeBalance"]

for columna in columnas_clave:
    minimo  = df[columna].min()
    maximo  = df[columna].max()
    media   = df[columna].mean().round(2)
    mediana = df[columna].median()
    print(f"\n{columna}:")
    print(f"  Min: {minimo}  |  Max: {maximo}  |  Media: {media}  |  Mediana: {mediana}")
```

Qué hace cada línea:
df.describe().T → ya lo conoces — estadísticas de todas las numéricas. El .T transpone la tabla para que sea más legible 😊
columnas_clave = [...] → lista con las columnas más importantes para el proyecto — las que más nos interesan para entender la retención de empleados
for columna in columnas_clave: → recorre cada columna de la lista
df[columna].min() → el valor mínimo de esa columna
df[columna].max() → el valor máximo
df[columna].mean().round(2) → la media redondeada a 2 decimales
df[columna].median() → la mediana — el valor del medio
print(f"\n{columna}:") → imprime el nombre de la columna con un salto de línea antes para que quede separado visualmente
